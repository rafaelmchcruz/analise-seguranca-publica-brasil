## Inspeção e tratamento da tabela de Mortes Violentas Intencionais (MVI)

Nesta etapa, serão selecionados e tratados os dados de Mortes Violentas Intencionais (MVI) por Unidade da Federação, referentes aos anos de 2023 e 2024. 

```python
# Definindo o caminho do arquivo original
arquivo = "../data/raw/anuario_seguranca_publica_2025.xlsx"
```

```python
# Carregando o arquivo Excel
excel = pd.ExcelFile(arquivo)

# Lista as abas disponíveis
excel.sheet_names
```

```python
tabelas_projeto = ['T01', 'T04', 'T13', 'T24']

for tabela in tabelas_projeto:
    print(tabela, "->", tabela in excel.sheet_names)
```

```python
# Carrega a Tabela 01 sem definir cabeçalho
df_t01 = pd.read_excel(arquivo, sheet_name='T01', header=None)

# Exibe as 15 primeiras linhas
df_t01.head(15)
```

```python
# Seleciona as linhas e colunas referentes ao MVI
mvi = df_t01.iloc[10:39, [0, 11, 12, 13, 14]].copy()

# Os primeiros registros
mvi.head()
```

```python
# Renomeia as colunas selecionadas
mvi.columns = [
    'uf',
    'mvi_2023',
    'mvi_2024',
    'taxa_2023',
    'taxa_2024'
]

mvi.head()
```

```python
# Removendo linhas totalmente vazias
mvi = mvi.dropna(how='all')

mvi.head()
```

```python
# Exibe os últimos registros
mvi.tail()
```

```python
# Verificando a quantidade de linhas e colunas
mvi.shape
```

```python
# Verificando os tipos das colunas
mvi.dtypes
```

```python
# Verifica os tipos dos valores da coluna
mvi['mvi_2023'].apply(type).value_counts()
```

```python
# Verifica os tipos dos valores da coluna
mvi['taxa_2023'].apply(type).value_counts()
```

```python
# Convertendo as ocorrências para valores numéricos
mvi['mvi_2023'] = pd.to_numeric(mvi['mvi_2023'])
mvi['mvi_2024'] = pd.to_numeric(mvi['mvi_2024']).astype(int)

# Convertendo as taxas para valores numéricos
mvi['taxa_2023'] = pd.to_numeric(mvi['taxa_2023'])
mvi['taxa_2024'] = pd.to_numeric(mvi['taxa_2024'])
```

```python
# Conferindo os tipos após a conversão
mvi.dtypes
```

```python
# Verificando valores nulos
mvi.isnull().sum()
```

```python
# Transformação para o formato: uf | ano | indicador | ocorrencias | taxa_100_mil.

# Cria a base de MVI para 2023
mvi_2023 = mvi[['uf', 'mvi_2023', 'taxa_2023']].copy()

# Padroniza os nomes das colunas
mvi_2023.columns = ['uf', 'ocorrencias', 'taxa_100_mil']

# Adiciona o ano e o indicador
mvi_2023['ano'] = 2023
mvi_2023['indicador'] = 'MVI'

# Cria a base de MVI para 2024
mvi_2024 = mvi[['uf', 'mvi_2024', 'taxa_2024']].copy()

# Padroniza os nomes das colunas
mvi_2024.columns = ['uf', 'ocorrencias', 'taxa_100_mil']

# Adiciona o ano e o indicador
mvi_2024['ano'] = 2024
mvi_2024['indicador'] = 'MVI'
```

```python
# Unindo os dados de 2023 e 2024
mvi_long = pd.concat([mvi_2023, mvi_2024], ignore_index=True)

# Definindo a ordem das colunas
mvi_long = mvi_long[
    ['uf', 'ano', 'indicador', 'ocorrencias', 'taxa_100_mil']
]

# Exibe o resultado da transformação
mvi_long.head()
```

```python
# Verificando as dimensões da base tratada
mvi_long.shape
```

```python
# Verificando os últimos registros
mvi_long.tail()
```

```python
# Convertendo contagens para inteiro
latrocinio['vitimas_2023'] = pd.to_numeric(
    latrocinio['vitimas_2023']
).astype(int)

latrocinio['vitimas_2024'] = pd.to_numeric(
    latrocinio['vitimas_2024']
).astype(int)

latrocinio['ocorrencias_2023'] = pd.to_numeric(
    latrocinio['ocorrencias_2023']
).astype(int)

latrocinio['ocorrencias_2024'] = pd.to_numeric(
    latrocinio['ocorrencias_2024']
).astype(int)

# Convertendo taxas para valores numéricos
latrocinio['taxa_vitimas_2023'] = pd.to_numeric(
    latrocinio['taxa_vitimas_2023']
)

latrocinio['taxa_vitimas_2024'] = pd.to_numeric(
    latrocinio['taxa_vitimas_2024']
)

latrocinio['taxa_ocorrencias_2023'] = pd.to_numeric(
    latrocinio['taxa_ocorrencias_2023']
)

latrocinio['taxa_ocorrencias_2024'] = pd.to_numeric(
    latrocinio['taxa_ocorrencias_2024']
)
```

```python
# Confererindo os tipos após conversão
latrocinio.dtypes
```

```python
# Transformando a tabela para o formato longo, preservando a diferença entre vítimas e ocorrências
# Dados de vítimas - 2023
latrocinio_vitimas_2023 = latrocinio[
    ['uf', 'vitimas_2023', 'taxa_vitimas_2023']
].copy()

latrocinio_vitimas_2023.columns = [
    'uf', 'ocorrencias', 'taxa_100_mil'
]

latrocinio_vitimas_2023['ano'] = 2023
latrocinio_vitimas_2023['indicador'] = 'Latrocínio - vítimas'


# Dados de vítimas - 2024
latrocinio_vitimas_2024 = latrocinio[
    ['uf', 'vitimas_2024', 'taxa_vitimas_2024']
].copy()

latrocinio_vitimas_2024.columns = [
    'uf', 'ocorrencias', 'taxa_100_mil'
]

latrocinio_vitimas_2024['ano'] = 2024
latrocinio_vitimas_2024['indicador'] = 'Latrocínio - vítimas'
```

```python
# Dados de ocorrências - 2023
latrocinio_ocorrencias_2023 = latrocinio[
    ['uf', 'ocorrencias_2023', 'taxa_ocorrencias_2023']
].copy()

latrocinio_ocorrencias_2023.columns = [
    'uf', 'ocorrencias', 'taxa_100_mil'
]

latrocinio_ocorrencias_2023['ano'] = 2023
latrocinio_ocorrencias_2023['indicador'] = 'Latrocínio - ocorrências'


# Dados de ocorrências - 2024
latrocinio_ocorrencias_2024 = latrocinio[
    ['uf', 'ocorrencias_2024', 'taxa_ocorrencias_2024']
].copy()

latrocinio_ocorrencias_2024.columns = [
    'uf', 'ocorrencias', 'taxa_100_mil'
]

latrocinio_ocorrencias_2024['ano'] = 2024
latrocinio_ocorrencias_2024['indicador'] = 'Latrocínio - ocorrências'
```

```python
# Juntando os dados de vítimas e ocorrências
latrocinio_long = pd.concat([
    latrocinio_vitimas_2023,
    latrocinio_vitimas_2024,
    latrocinio_ocorrencias_2023,
    latrocinio_ocorrencias_2024
], ignore_index=True)

# Padronizando a ordem das colunas
latrocinio_long = latrocinio_long[
    ['uf', 'ano', 'indicador', 'ocorrencias', 'taxa_100_mil']
]

# Exibindo resultado
latrocinio_long.head()
```
